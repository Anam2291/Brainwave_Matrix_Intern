## **Stemming**
#### *Stemming is the process of reducing a word to its Root Word*
**example: actor, acting, actress = act**

```python
# Initialize the stemmer
port_stem = PorterStemmer()

def stemming(content):
    # Remove non-alphabet characters
    stemmed_content = re.sub('[^a-zA-Z]', ' ', content)
    # Convert to lowercase
    stemmed_content = stemmed_content.lower()
    # Split into words
    stemmed_content = stemmed_content.split()
    # Apply stemming and remove stopwords
    stemmed_content = [port_stem.stem(word) for word in stemmed_content if word not in stopwords.words('english')]
    # Join back into a single string
    stemmed_content = ' '.join(stemmed_content)
    return stemmed_content
```

```python
# Apply the function to the DataFrame
twitter_data['stemmed_content'] = twitter_data['tweet'].apply(stemming)
```

### Separating the data and label

```python
X = twitter_data['stemmed_content'].values
y = twitter_data['label'].values
```

```python
print(X)
```

```python
print(y)
```

### Splitting the data to training and test data

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, stratify= y, random_state=2)
# stratify, in order to tell train_test split we want equal propotion of the two classes which is '0'& '1'both in training & test data.
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
print(X_test)
```

## Feature Extraction
**Converting the Textual data to Numerical data**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize the vectorizer
vectorizer = TfidfVectorizer()

# Fit the vectorizer on the training data and transform it
X_train = vectorizer.fit_transform(X_train)

# Transform the test data using the same vectorizer
X_test = vectorizer.transform(X_test)
```

```python
print(X_train)
```

```python
print(X_test)
```

### Training the 'Machine Learning Model'
**Logistic Regression**\
It is Classification Model to classify the data points into different classes

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train, y_train)
```

### Model Evaluation
**Accuracy Score**

```python
from sklearn.metrics import accuracy_score
# Accuracy score on the training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(y_train, X_train_prediction)*100
# we are comparing the true values(y_train)and the values predicted by our model
```

```python
print('Accuracy_score on the training data :', training_data_accuracy)
```

```python

# Accuracy score on the test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(y_test, X_test_prediction)*100
```

```python
print('Accuracy_score on the test data :', test_data_accuracy)
```

**Logistic Regression model demonstrates excellent performance, with 95.12% test accuracy and 95.40% training accuracy.** \
**The small difference between the two indicates no significant overfitting, meaning the model generalizes well to new data. This high accuracy suggests the model effectively captures patterns in the dataset.**

**Model Accuracy = 95.11%**

```python
from sklearn.metrics import classification_report

# Get detailed classification report
print(classification_report(y_test, X_test_prediction))
```

#### **-Class 0 (negative sentiment):** The model performs exceptionally well, with 95% precision and 100% recall, meaning it correctly identifies almost all negative sentiments.

#### **-Class 1 (positive sentiment):** The model has high precision (91%), but low recall (33%), meaning it misses many actual positive sentiment tweets (high false negatives).

```python
import numpy as np 
fnl = pd.DataFrame(np.c_[y_test, X_test_prediction], columns = ['Actual','Predicted'])
fnl
```

#### **np.c_[]:** This is a NumPy function that concatenates arrays along the second axis (columns). In this case, it combines the true labels (y_test) and the predicted labels (X_test_prediction) into a single 2D array.

#### **pd.DataFrame():** This converts the 2D array into a Pandas DataFrame, which is a tabular data structure that can hold different types of data.

#### **columns = ['Actual', 'Predicted']:** This specifies the column names for the DataFrame. The first column will be labeled 'Actual' (representing the true labels), and the second column will be labeled 'Predicted' (representing the model's predictions).
#### **The resulting DataFrame fnl contains two columns:** one for the actual sentiment labels from the test set and another for the predicted sentiment labels, allowing for easy comparison of model performance.**

## Summary:
**The model excels at identifying negative sentiments (Class 0), achieving 95% precision and 100% recall. \
However, it struggles with positive sentiments (Class 1), with high precision (91%) but low recall (33%), indicating many missed actual positive tweets.**
